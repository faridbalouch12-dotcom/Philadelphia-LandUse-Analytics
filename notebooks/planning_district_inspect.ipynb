```python
import geopandas as gpd
import pandas as pd
from urllib.parse import quote
from datetime import datetime, timedelta

BASE_URL = "https://services.arcgis.com/fLeGjb7u4uXqeF9q/arcgis/rest/services/Permits/FeatureServer/0/query"

cutoff_date = (datetime.now() - timedelta(days=8 * 365)).strftime("%Y-%m-%d")

WHERE_CLAUSE = quote(
    f"systemofrecord = 'ECLIPSE' "
    f"AND UPPER(status) = 'ISSUED' "
    f"AND permitissuedate >= DATE '{cutoff_date}'"
)

def fetch_all_features(batch_size=1000):
    all_gdfs = []
    offset = 0

    while True:
        url = (
            f"{BASE_URL}"
            f"?where={WHERE_CLAUSE}"
            f"&outFields=*"
            f"&f=geojson"
            f"&resultOffset={offset}"
            f"&resultRecordCount={batch_size}"
        )

        print(f"Fetching rows {offset} to {offset + batch_size}...")
        batch = gpd.read_file(url)

        if batch.empty or len(batch) < batch_size:
            if not batch.empty:
                all_gdfs.append(batch)
            break

        all_gdfs.append(batch)
        offset += batch_size

    return gpd.GeoDataFrame(pd.concat(all_gdfs, ignore_index=True), geometry="geometry", crs="EPSG:4326")


permits = fetch_all_features()
print(f"Total rows fetched: {len(permits)}")
```

```python
permits["permitissuedate"] = pd.to_datetime(permits["permitissuedate"], unit="ms").dt.date
```
