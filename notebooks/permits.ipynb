```python
import geopandas as gpd
import pandas as pd

BASE_URL = "https://hub.arcgis.com/api/v3/datasets/7778f8535c744f9b9000c8be95bb6d84_0/downloads/data"

def fetch_all_features(batch_size=1000):
    all_gdfs = []
    offset = 0

    while True:
        url = (
            f"{BASE_URL}"
            f"?format=geojson"
            f"&spatialRefId=4326"
            f"&where=1%3D1"
            f"&resultOffset={offset}"
            f"&resultRecordCount={batch_size}"
        )

        print(f"Fetching rows {offset} to {offset + batch_size}...")
        batch = gpd.read_file(url)

        if batch.empty:
            print("No more rows. Done.")
            break

        all_gdfs.append(batch)
        offset += batch_size

        # If the batch returned fewer rows than requested, we've hit the end
        if len(batch) < batch_size:
            print(f"Last batch had {len(batch)} rows. Done.")
            break

    return pd.concat(all_gdfs, ignore_index=True)


permits = fetch_all_features(batch_size=1000)
print(f"Total rows fetched: {len(permits)}")
```
